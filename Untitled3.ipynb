```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=pd.read_csv('/content/advertising (1).csv')
```

```python
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
plt.figure(figsize=(10,10))
sns.pairplot(data,x_vars=['TV','Radio','Newspaper'],y_vars=['Sales'],height=7,kind='scatter')
plt.xlabel('advertisement')
plt.ylabel('sales')
plt.show()
```

```python
x=data[['TV','Radio','Newspaper']]
x.head()
```

```python
y=data[['Sales']]
y.head()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.7,random_state=100)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
lr.fit(x_train,y_train)
```

```python
y_train_predict=lr.predict(x_train)
```

```python
from os import error
from sklearn.metrics import mean_squared_error
mse_train=mean_squared_error(y_train,y_train_predict)
```

```python
from sklearn.metrics import r2_score
r2_train=r2_score(y_train,y_train_predict)
```

```python
print('Mse',mse_train)
print('accuracy',r2_train)
```

```python
c = [i for i in range (1,len(y_test)+1)]
y_predict = lr.predict(x_test)
plt.plot(c,y_test['Sales'],color='r',linestyle='-', label='Actual Sales')
plt.plot(c,y_predict,color='b',linestyle='-', label='Predicted Sales')
plt.xlabel('Data Point Index')
plt.ylabel('Sales')
plt.title('Prediction')
plt.legend()
plt.show()
```

```python
c=[i for i in range(1,len(y_test)+1)]
plt.plot(c,y_test-y_predict,color='green',linestyle='-')
plt.xlabel('Advertisement')
plt.ylabel('Error')
plt.title('Error value')
plt.show()
```

```python
from sklearn.metrics import mean_squared_error,r2_score
```
